```python
import pandas as pd

df = pd.read_csv(r"C:\online_retail\data\raw\online_retail_II.csv")
```

```python
df["Amount"] = df["Quantity"] * df["Price"]

df[["Invoice", "StockCode", "Description", "Quantity", "Price", "Amount"]].head(20)
```

```python
df["Amount"].describe()
```

```python
df.nlargest(20, "Amount")[
    ["Invoice", "StockCode", "Description",
     "Quantity", "Price", "Amount", "Customer ID", "Country"]
]
```

```python
df.nsmallest(20, "Amount")[
    ["Invoice", "StockCode", "Description",
     "Quantity", "Price", "Amount", "Customer ID", "Country"]
]
```

```python
df.assign(
    InvoiceType=df["Invoice"].astype(str).str[0]
).query(
    "Amount < 0"
).groupby("InvoiceType")["Amount"].agg(
    ["count", "sum", "min", "max"]
)
```

```python
df["Amount"].sum()
```

```python
df["Amount"].gt(0).sum()
df["Amount"].lt(0).sum()
```

```python
df.groupby(
    df["Invoice"].astype(str).str[0]
)["Amount"].agg(["count", "sum", "mean", "min", "max"])
```
