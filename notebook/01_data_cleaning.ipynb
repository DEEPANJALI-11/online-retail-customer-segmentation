```python
invoice_country = (
    df.assign(InvoiceType=df["Invoice"].astype(str).str[0])
      .query("InvoiceType in ['C', 'A']")
      .groupby(["InvoiceType", "Country"])["Invoice"]
      .nunique()
      .reset_index(name="Unique_Invoices")
      .sort_values(["InvoiceType", "Unique_Invoices"], ascending=[True, False])
)

invoice_country
```

```python
negative_price = df[df["Price"] < 0]

negative_price.groupby(
    negative_price["Invoice"].astype(str).str[0]
).size()
```

```python
price_by_invoice_type = (
    df.assign(
        InvoiceType=df["Invoice"].astype(str).str[0]
    )
    .groupby("InvoiceType")
    .agg(
        Total_Rows=("Price", "size"),
        Zero_Price=("Price", lambda x: (x == 0).sum()),
        Negative_Price=("Price", lambda x: (x < 0).sum()),
        Positive_Price=("Price", lambda x: (x > 0).sum())
    )
    .reset_index()
)

price_by_invoice_type
```

```python
c_negative_price = df[
    (df["Invoice"].astype(str).str.startswith("C")) &
    (df["Price"] < 0)
]

print("Number of C invoices with negative price:", len(c_negative_price))
```
