```python
import pandas as pd

df = pd.read_csv(r"C:\online_retail\data\raw\online_retail_II.csv")
```

```python
df.assign(
    InvoiceType=df["Invoice"].astype(str).str[0]
).query(
    "InvoiceType in ['C', 'A']"
).groupby("InvoiceType")["Country"].nunique()
```

```python
price_by_invoice_type = (
    df.assign(
        InvoiceType=df["Invoice"].astype(str).str[0]
    )
    .groupby("InvoiceType")
    .agg(
        Total_Rows=("Price", "size"),
        Zero_Price=("Price", lambda x: (x == 0).sum()),
        Negative_Price=("Price", lambda x: (x < 0).sum()),
        Positive_Price=("Price", lambda x: (x > 0).sum())
    )
    .reset_index()
)

price_by_invoice_type
```

```python
c_data = df[
    df["Invoice"].astype(str).str.startswith("C")
].copy()

c_data["Amount"] = c_data["Quantity"] * c_data["Price"]

c_data[["Invoice", "Quantity", "Price", "Amount"]].head(20)
```

```python
print("Negative quantity:", (c_data["Quantity"] < 0).sum())
print("Zero quantity:", (c_data["Quantity"] == 0).sum())
print("Positive quantity:", (c_data["Quantity"] > 0).sum())
```
